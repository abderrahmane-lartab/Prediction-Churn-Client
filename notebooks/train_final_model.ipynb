#### Train final model
This notebook trains the final churn model using the clustered train/test feature sets.


```python
import joblib
from xgboost import XGBClassifier

X_train = joblib.load("../data/processed/X_train.pkl")
y_train = joblib.load("../data/processed/y_train.pkl")



model = XGBClassifier(

        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_estimators=200,
        learning_rate=0.01,
        colsample_bytree= 1.0,
        max_depth= 3,
        scale_pos_weight=6,
        subsample= 1.0
    )
model.fit(X_train, y_train)



joblib.dump(model, "../data/processed/models/final_model.pkl")
print("Model saved successfully.")
```
